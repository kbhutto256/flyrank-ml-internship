```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
import pandas as pd
import os

# Ensure the 'data/raw' directory exists
os.makedirs('data/raw', exist_ok=True)

# Define the URL for the raw data file
data_url = 'https://raw.githubusercontent.com/kbhutto256/flyrank-ml-internship/main/data/raw/content_refresh_anonymized.csv'
# Define the local path where the file should be saved
local_file_path = 'data/raw/content_refresh_anonymized.csv'

# Check if the file already exists to avoid re-downloading
if not os.path.exists(local_file_path):
    print(f"Downloading {local_file_path}...")
    !wget -q -O {local_file_path} {data_url}
    print("Download complete.")
else:
    print(f"File {local_file_path} already exists. Skipping download.")

df = pd.read_csv(local_file_path)

# Number 1: how much of the inventory is currently trending down?
declining_pct = (df['trend_direction'] == 'down').mean()
print(f"% of pages currently trending down: {declining_pct:.1%}")

# Number 2: how many declining pages actually have real demand worth prioritizing?
visible_declining = df[(df['trend_direction'] == 'down') & (df['impressions_90d'] >= 100)]
print(f"Declining pages with >=100 impressions/90d: {len(visible_declining)} of {len(df)} total")

# Number 3: baseline vs. learned model gap already proven in the starter pipeline
print("Starter pipeline Precision@50 — baseline rule: 0.240, random forest: 0.740")
```
